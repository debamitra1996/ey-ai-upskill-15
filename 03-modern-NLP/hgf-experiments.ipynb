```python
from transformers import pipeline
```

```python
model = pipeline("sentiment-analysis")
```

```python
model("Today is a great day for learning AI")
```

```python
model("The movie was not very good, but OK")
```

### Tasks supported by pipeline()

['any-to-any', 'audio-classification', 'automatic-speech-recognition', 'depth-estimation', 'document-question-answering', 'feature-extraction', 'fill-mask', 'image-classification', 'image-feature-extraction', 'image-segmentation', 'image-text-to-text', 'keypoint-matching', 'mask-generation', 'ner', 'object-detection', 'sentiment-analysis', 'table-question-answering', 'text-classification', 'text-generation', 'text-to-audio', 'text-to-speech', 'token-classification', 'video-classification', 'zero-shot-audio-classification', 'zero-shot-classification', 'zero-shot-image-classification', 'zero-shot-object-detection']

### Zero Shot Classification

```python
classifier = pipeline( 'zero-shot-classification' )
```

```python
classifier("It was great learning experience in the course", candidate_labels=["sports", "education", "politics", "business"])
```

```python
classifier("India won the cricket match by 5 wickets", candidate_labels=["sports", "education", "politics", "business"])
```

### Text Generation

```python
model = "HuggingFaceTB/SmolLM2-360M"
```

```python
textgen = pipeline("text-generation", model=model)
```

```python
text = textgen("Define Quantum Computing in 100 words")
```

```python
print(text[0]["generated_text"])
```

### Image Classification

```python
imgclassifier = pipeline(
    task="image-classification",
    model="google/vit-base-patch16-224"
)
```

```python
link = r"https://img.magnific.com/free-photo/portrait-white-man-isolated_53876-40306.jpg?semt=ais_hybrid&w=740&q=80"
imgclassifier(link)
```
